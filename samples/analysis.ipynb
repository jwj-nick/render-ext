# MAC Array Utilization Study

NVDLA CMAC 배열의 **활용률**을 레이어별로 살펴본다.

```mermaid
flowchart LR
    W[weights] --> M[MAC array]
    A[activations] --> M
    M --> ACC[accumulator]
```


````python
import numpy as np

layers = ["conv1", "conv2", "conv3", "fc1", "fc2"]
util = np.array([0.87, 0.81, 0.74, 0.31, 0.22])
print(f"layers = {len(layers)}, macs = 1024")
print(f"peak utilization = {util.max():.2f}")
````

## Observation

FC 레이어에서 활용률이 급락한다 — 가중치 재사용이 없기 때문.

````python
import pandas as pd
pd.DataFrame({"layer": layers[:1] + layers[3:4], "util": [0.87, 0.31]})
````
